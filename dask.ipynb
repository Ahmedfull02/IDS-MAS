```python
import dask_ml.datasets as dask_datasets
from dask_ml.cluster import KMeans

# Load a distributed dataset
X, y = dask_datasets.make_classification(n_samples=100000, chunks=10000)

# Instantiate and fit a distributed K-means clustering model
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

# Get the cluster labels for the dataset
labels = kmeans.labels_
print("Cluster Labels:", labels.compute())
```
