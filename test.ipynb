```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Define labels and matrix data
labels = ["BENIGN", "Botnet", "DoS", "FTP-Patator", "Port Scanning"]

confusion_matrix_data = [
    [0.999236, 0.000080, 0.000557, 0.000000, 0.000127], # BENIGN
    [0.007292, 0.991667, 0.000000, 0.000000, 0.001042], # Botnet
    [0.001687, 0.000000, 0.998313, 0.000000, 0.000000], # DoS
    [0.006274, 0.000000, 0.000000, 0.993726, 0.000000], # FTP-Patator
    [0.001471, 0.000000, 0.000285, 0.000000, 0.998243]  # Port Scanning
]
# Convert to a DataFrame for easier labeling with Seaborn
df_cm = pd.DataFrame(confusion_matrix_data, index=labels, columns=labels)

# Set up the matplotlib figure size
plt.figure(figsize=(12, 8))

# Create the heatmap
# annot=True: write the data value in each cell
# fmt='.6f': format as float with 6 decimal places (matches original image)
# cmap='Blues': use the blue color map
sns.heatmap(df_cm, annot=True, fmt='.6f', cmap='Blues', annot_kws={"size": 16})

# Add titles and labels
plt.title('Confusion Matrix Normalised')
plt.ylabel('True label')
plt.xlabel('Predicted label')

# Display the plot
plt.show()
```
